```python
import hmplot as hm

# A brushed Group's rows read back to Python (TODO 145.1): await totals.read() after a brush.
sales = hm.Table({"region": ["north", "south", "east", "west"] * 25,
                  "amount": [float((i * 37) % 100) for i in range(100)],
                  "units": [float(i % 13) for i in range(100)]})
with hm.Scene() as scene:
    with hm.Data(sales):
        with hm.Row():
            with hm.Plot(height=240, title="Orders", x={"domain": [0, 100]}, y={"domain": [0, 13]}):
                hm.Points(x="amount", y="units", brush=True, respond="brushed")
            with hm.Filter(where="@brushed", inactive="all"):
                with hm.Group(by="region", columns={"total": hm.sum("amount")}) as totals:
                    with hm.Plot(height=240, title="Totals"):
                        hm.Points(x="region", y="total", size=12)
scene
```
