# Lollipop


```python
import hmplot as hm

# Values on both sides of the baseline get a stem; theme.marks.Lollipop styles the parts.
table = hm.Table({"category": ["Alder", "Birch", "Cedar", "Douglas fir", "Elm", "Hemlock"],
                  "change": [18, 31, -12, 44, 26, -9]})
with hm.Plot(data=table, axes2d=True, y={"domain": [-20, 50], "tick_count": 3}) as plot:
    hm.charts.Lollipop(x="category", y="change", fill="#547ba2", size=7, stem={"width": 2})
plot
```
