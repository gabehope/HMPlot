# Dataframes and categories


```python
import hmplot as hm
import numpy as np
import pandas as pd

frame = pd.DataFrame(
    {
        "length": [5.1, 4.9, 6.2],
        "width": [3.5, 3.0, 3.4],
        "species": ["setosa", "setosa", "virginica"],
    }
)
with hm.Data(frame) as view:
    with hm.Plot(axes2d=True):
        # A channel takes a column by name, or the column itself: a Series titles its axis with
        # its name.
        points = hm.Points(x="length", y=frame["width"], fill="species", size=3.5)
        hm.Legend(target=points, channel="fill", label="Species")
view
```
