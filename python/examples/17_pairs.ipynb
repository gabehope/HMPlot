# Scatterplot matrices from shared columns


```python
import hmplot as hm
import numpy as np
t = np.linspace(0, 6, 32)
with hm.Scene() as scene:
    with hm.Data({"time": t, "sine": np.sin(t), "cosine": np.cos(t)}):
        with hm.Facet(pairs=["time", "sine", "cosine"]):
            with hm.Plot(height=220, axes2d=True):
                hm.Points(x="x", y="y", size=2.5)
scene
```

```python
hm.visualize_tree(scene)
```
