# Legends for integer, datetime, and shared-scale columns
A colorbar and a size ramp over integer columns label the data's own values; over a datetime64 column they label dates. A scale's `label` titles the colorbar shared by its marks. The browser test `tests/browser/legends-python.spec.js` renders this scene from `static/test/notebook-48_legend_columns.json`.

```python
"""Legend ticks for integer and datetime columns.

Integer columns keep their values in the colorbar and the size ramp (fill 100..180 labels
100..180), datetime64 columns label their ticks as dates, not raw milliseconds, and a scale's
label titles the colorbar its marks share.
"""
import numpy as np

import hmplot as hm

n = 9
data = {
    "x": np.linspace(5.0, 85.0, n),
    "y": np.linspace(10.0, 90.0, n),
    "count": np.arange(100, 181, 10, dtype=np.int64),
    "mass": np.array([10, 50, 90, 130, 170, 210, 250, 290, 310], dtype=np.int64),
    "when": np.datetime64("2024-01-01T12:00") + np.arange(n) * np.timedelta64(45, "D"),
}

with hm.Scene() as scene:
    with hm.Row(wrap=True, min_width=320):
        with hm.Plot(data=data, height=300, axes2d=True):
            counts = hm.Points(x="x", y="y", fill="count", size="mass",
                                      size_scale=hm.continuous(range=[4, 20]), label="Counts")
            hm.Legend(target=counts, channel="fill", label="Count", position="right")
            hm.Legend(target=counts, channel="size", label="Mass", position="bottom")
        with hm.Plot(data=data, height=300, axes2d=True):
            dated = hm.Points(x="x", y="y", fill="when", size="when",
                                     size_scale=hm.continuous(range=[4, 20]), label="Dated")
            hm.Legend(target=dated, channel="fill", label="When", position="right")
            hm.Legend(target=dated, channel="size", label="When (size)", position="bottom")
        # Two marks share a labeled scale (one colorbar, titled by the scale); the third
        # mark's automatic scale gets its own colorbar, titled by the mark's label.
        shared = hm.continuous(domain=[0, 100], colormap="Magma", label="Shared value")
        with hm.Plot(data=data, height=300, axes2d=True):
            hm.Points(x="x", y="y", fill="x", fill_scale=shared, size=4)
            hm.Points(x="y", y="x", fill="y", fill_scale=shared, size=2.5)
            hm.Points(x="x", y="x", fill="y", size=2, label="Diagonal")
            hm.Legend(channel="fill", position="right")
scene
```
