# Point labels
Labels are separate GPU Text marks sharing the points’ data coordinates. The small horizontal offset is in data units.

```python
import hmplot as hm

observations = [
    {'name': 'Alder', 'x': 2, 'y': 8},
    {'name': 'Birch', 'x': 7.5, 'y': 8.3},
    {'name': 'Cedar', 'x': 1.5, 'y': 6},
    {'name': 'Maple', 'x': 7, 'y': 6.2},
    {'name': 'Aspen', 'x': 2.4, 'y': 4},
    {'name': 'Beech', 'x': 7.8, 'y': 4.2},
    {'name': 'Hazel', 'x': 1.8, 'y': 2},
    {'name': 'Willow', 'x': 7.2, 'y': 1.8},
]
labels = []
with hm.Scene() as scene:
    with hm.Plot(data=observations, height=440, axes2d=True, aspect="equal",
                 x={'domain': [0, 12], 'label': 'X'},
                 y={'domain': [0, 10], 'label': 'Y'}) as plot:
        hm.Points(x='x', y='y', size=8,
                         fill='#547ba2', outline_color='#f4f2ed', outline_width=0.7)
        for observation in observations:
            labels.append(hm.Text3D(
                text=observation['name'],
                position=[observation['x'] + 0.45, observation['y'], 0],
                align='left', size=4, color='#30343b',
                name=observation['name']))
scene
```

```python
# Change these values and rerun this cell to update the existing labels.
show_labels = True
label_size = 5.0
with scene.batch_update():
    for label in labels:
        label.update(visible=show_labels, size=label_size)
await scene.when_settled()
```
