```python
import hmplot as hm

# One Filter > Group pipeline, computed once and drawn by two plots through references (TODO 142).
sales = hm.Table({"region": ["north", "south", "east", "west"] * 25,
                  "amount": [float((i * 37) % 100) for i in range(100)]})
with hm.Scene() as scene:
    with hm.Data(sales):
        with hm.Filter(where=hm.col.amount > 20):
            totals = hm.Group(by="region", columns={"orders": hm.count(), "total": hm.sum("amount")})
    with hm.Row():
        with hm.Plot(data=totals, height=240, title="Orders"):
            hm.Points(x="region", y="orders", size=12, select=True)
        with hm.Plot(height=240, title="Totals"):
            with hm.Data(value=totals):
                hm.Points(x="region", y="total", size=12, respond="selected")
scene
```
