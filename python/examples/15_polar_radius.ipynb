# Polar coordinates use angle and radius


```python
import hmplot as hm
import numpy as np
angle = np.linspace(0, 360, 49)
data = {"angle": angle, "radius": 2 + 0.6 * np.cos(np.deg2rad(angle * 4))}
with hm.Plot(data=data, coordinates="polar", base_radius=0, radial_scale=1, x={"domain": [0, 360]}, y={"domain": [0, 3]}, axes2d=True) as plot:
    hm.Line(x="angle", y="radius", width=2)
    hm.Points(x="angle", y="radius", size=2.5)
plot
```
