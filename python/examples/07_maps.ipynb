# Globe maps and GeoJSON


```python
import hmplot as hm
import numpy as np

regions = {
    "type": "FeatureCollection",
    "features": [
        {
            "type": "Feature",
            "id": "west",
            "properties": {"group": "A", "value": 0.15},
            "geometry": {
                "type": "Polygon",
                "coordinates": [[[-50, 0], [-20, 0], [-20, 30], [-50, 30], [-50, 0]]],
            },
        },
        {
            "type": "Feature",
            "id": "east",
            "properties": {"group": "B", "value": 0.3},
            "geometry": {
                "type": "Polygon",
                "coordinates": [[[10, -20], [40, -20], [40, 10], [10, 10], [10, -20]]],
            },
        },
    ],
}
with hm.Plot(coordinates="spherical", type="3d-y-up", overview=True, lights=[hm.light('ambient', intensity=1), hm.light('directional', position=[3, 4, 5], intensity=2)]) as globe:
    regions_mark = hm.GeoJSON(data=regions, fill="group", height="value", lit=True)
    hm.Legend(target=regions_mark, channel="fill")
globe
```
