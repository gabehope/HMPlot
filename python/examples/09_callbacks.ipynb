# Callbacks and export

Click a point and watch the status below the plot update. Callback `print()` output is not reliably attached to a finished notebook cell, so this example updates a persistent widget instead. Run the export cell after the plot appears.


```python
import hmplot as hm
import ipywidgets as widgets
from IPython.display import display

print("HMPlot", hm.__version__, "from", hm.__file__)
with hm.Plot(axes2d=True, height=420,
             x={"domain": [-0.3, 2.3]}, y={"domain": [-0.2, 1.2]}) as plot:
    points = hm.Points(position=[[0, 0], [1, 0.5], [2, 1]],
                              size=40, fill="#547ba2")
selected = []
status = widgets.HTML(value="Click a point to inspect it.")

def inspect(event):
    selected.append(event.source_index)
    xy = event.data_coordinates or {}
    status.value = (f"Selected row <b>{event.source_index}</b> at "
                    f"({xy['x']:.2f}, {xy['y']:.2f}). Clicks: {len(selected)}")

points.on_click(inspect)
display(status, plot)
```

```python
png = await plot.export_png("points.png")
print("Saved points.png:", len(png), "bytes")
```
