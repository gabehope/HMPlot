# Sunburst


```python
import hmplot as hm

rows = [
    {"startAngle": 0, "endAngle": 360, "innerRadius": 0, "outerRadius": .32, "category": "Root", "name": "All"},
    {"startAngle": 0, "endAngle": 150, "innerRadius": .35, "outerRadius": .66, "category": "A", "name": "A"},
    {"startAngle": 150, "endAngle": 360, "innerRadius": .35, "outerRadius": .66, "category": "B", "name": "B"},
]
for row in rows:
    row["labelAngle"] = (row["startAngle"] + row["endAngle"]) / 2
    row["labelRadius"] = (row["innerRadius"] + row["outerRadius"]) / 2
with hm.Plot(data=rows, coordinates="polar", base_radius=0, radial_scale=1, x={"domain": [0, 360]}, y={"domain": [0, 1]}) as plot:
    sectors = hm.Rect(x0="startAngle", x1="endAngle", y0="innerRadius", y1="outerRadius", fill="category")
    hm.Text(x="labelAngle", y="labelRadius", text="name", color="#f4f2ed")
    hm.Legend(target=sectors, channel="fill")
plot
```

```python
hm.visualize_tree(plot)
```
