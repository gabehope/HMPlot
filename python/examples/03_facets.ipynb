# Facets and inputs


```python
import hmplot as hm
import numpy as np

rng = np.random.default_rng(42)
with hm.Scene() as scene:
    # One input, read by the histogram in every cell.
    with hm.Row(wrap=True, min_width=0):
        hm.Slider(name="bins", min=4, max=40, step=1, value=12, label="Bins")
    with hm.Data(
        {"x": rng.normal(size=120), "group": ["A"] * 40 + ["B"] * 40 + ["C"] * 40}
    ):
        with hm.Facet(column="group", wrap=3):
            with hm.Plot(axes2d=True, height=240):
                with hm.stats.Bin(x="x", bins=hm.input.bins):
                    hm.Rect(x0="x0", x1="x1", y0=0, y1="count")
scene
```

```python
hm.visualize_tree(scene, split_facet=True)
```
