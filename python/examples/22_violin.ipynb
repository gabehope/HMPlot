# Violin


```python
import hmplot as hm
import numpy as np

groups = ["A", "B", "C"]
table = hm.Table({"group": [groups[i % 3] for i in range(240)], "value": np.sin(np.arange(240) * 2.17) + np.arange(240) % 3})
with hm.Plot(x={"scale": hm.categorical(domain=groups)}, data=table, axes2d=True) as plot:
    hm.charts.Violin(x="group", y="value", fill="group")
plot
```
