# Data shortcuts and twin axes

```python
import hmplot as hm
import numpy as np
hour = np.arange(25, dtype=np.float32)
data = hm.Table({'hour': hour, 'temperature': 15 + 8*np.sin(hour/24*np.pi*2), 'rainfall': np.maximum(0, 12*np.cos(hour/5))})
with hm.Plot(data=data, height=420) as view:
    with hm.Frame(name='Temperature', x={'domain':[0,24], 'label':'Hour'}, y={'domain':[5,25], 'label':'Temperature'}):
        hm.Line(x='hour', y='temperature', color='#547ba2')
    with hm.Frame(name='Rainfall', x={'domain':[0,24], 'visible':False}, y={'domain':[0,15], 'side':'right', 'label':'Rainfall'}):
        hm.Line(x='hour', y='rainfall', color='#cf8955')
view
```
