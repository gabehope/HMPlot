# Histograms and facet settings

```python
import hmplot as hm
import numpy as np
rng = np.random.default_rng(42)
data = {'value': rng.normal(size=120), 'group': ['A','B','C']*40}
with hm.Facet(data=data, column='group', wrap=3) as facets:
    with hm.Plot(axes2d=True, height=260):
        hm.charts.Histogram(x='value', bins=12, normalization='density', fill='#547ba2')
facets
```
