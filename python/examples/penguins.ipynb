```python
import pandas as pd
import hmplot as hm
```

```python
data = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv")
```

```python
data
```

```python
hm.Points(data=data, x='bill_length_mm', y='bill_depth_mm', fill='species')
```

```python
with hm.Plot(data=data, axes2d=True) as plot:
    hm.Points(x='bill_length_mm', y='bill_depth_mm', fill='species', animation=False, hover=True, size=hm.where('@hovered', 16, 12))

plot
```

```python
with hm.Facet(data=data, column='island', link='xy') as plot:
    with hm.Plot(axes2d=True):
        hm.Points(x='bill_length_mm', y='bill_depth_mm', fill='species', hover=True, size=hm.where('@hovered', 9, 6))
plot
```

```python
with hm.Facet(data=data, column='island', link='xy') as plot:
    with hm.Plot(axes2d=True):
        hm.Points(x='bill_length_mm', y='bill_depth_mm', fill='species')
        
        with hm.Sample(size=5):
            hm.Text(text='species', x='bill_length_mm', y='bill_depth_mm', color='species')
plot
```

```python
hm.visualize_tree(plot, split_facet=True)
```
