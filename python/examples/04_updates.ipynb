# Live NumPy updates

Run the setup cell once, then rerun the update cell to animate new positions.


```python
import hmplot as hm
import numpy as np

rng = np.random.default_rng(42)
with hm.Plot(x={"domain": [-3, 3]}, y={"domain": [-3, 3]}, axes2d=True) as plot:
    points = hm.Points(
        position=rng.normal(size=(50, 2)), animation=True, size=5, hover=True, respond='hovered'
    )
plot
```

```python
points.update(position=rng.normal(size=(50, 2)))
await plot.when_animations_finished(timeout=30)
print('Update complete')
```

## Optional export

This saves the current displayed view after the animation finishes.


```python
await plot.export_png('live-updates.png', scale=2, timeout=30)
print('Saved live-updates.png')
```
