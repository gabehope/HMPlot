```python
import hmplot as hm

# A drill-down Treemap built in Python (TODO 145.2-145.3, 145.7): a click on a rectangle writes the
# input "root" (root.set("id")), which the Treemap reads, a Select of that name shows, and the Top
# button resets (root.reset()).
root = hm.input("root", default="all")
edges = {"source": ["all", "all", "a", "a", "b", "b"], "target": ["a", "b", "a1", "a2", "b1", "b2"]}
nodes = {"id": ["all", "a", "b", "a1", "a2", "b1", "b2"], "weight": [0, 0, 0, 1, 2, 3, 4]}
with hm.Scene() as scene:
    hm.Select(name="root", options=["all", "a", "b"])
    hm.Button(label="Top", set=root.reset())
    with hm.Plot(height=260, x={"domain": [0, 1]}, y={"domain": [0, 1]}):
        with hm.GraphData(nodes=nodes, edges=edges, source="source", target="target"):
            with hm.Treemap(root=root, value="weight", levels=1, padding=0.02):
                with hm.Nodes() as shown:
                    hm.Rect(x0="x0", x1="x1", y0="y0", y1="y1", fill="#4a6f8f", select=root.set("id"))
scene
```
