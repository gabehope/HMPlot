# Shared mappings and pixel-sized labels


```python
import hmplot as hm
observations = {"x": [1, 2, 4], "y": [1, 3, 2], "name": ["Alder", "Birch", "Cedar"]}
with hm.Scene() as scene:
    with hm.Plot(data=observations, axes2d=True, x={"domain": [0, 6]}, y={"domain": [0, 4]}):
        with hm.Mapping(x="x", y="y", opacity=0.9):
            hm.Points(size=4, fill="#547ba2")
            hm.Text(text="name", size=14, dx=8, align="left")
scene
```
