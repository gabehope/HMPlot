# Deterministic rooted graph layout


```python
import hmplot as hm
edges = {"source": ["Root", "Root", "Left", "Right"], "target": ["Left", "Right", "Leaf A", "Leaf B"]}
with hm.Scene() as scene:
    with hm.Data(edges):
        with hm.GraphData(source="source", target="target"):
            with hm.Tree(root="Root"):
                with hm.Plot(x={"domain": [-2, 2]}, y={"domain": [-3, 1]}):
                    with hm.Edges():
                        hm.charts.Arrows(start="start", end="end", curve="link-vertical", width=1)
                    with hm.Nodes():
                        hm.Points(position="position", size=4)
                        hm.Text(x="x", y="y", text="id", dy=-14)
scene
```
