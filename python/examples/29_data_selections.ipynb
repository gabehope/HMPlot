# Data-owned selections and brushing
Click or Shift-drag in the first plot. The second plot reads the committed `@selected` virtual column.


```python
import math
import hmplot as hm

observations = hm.Table({
    'id': [f'sample-{i}' for i in range(30)],
    'x': [i / 29 for i in range(30)],
    'y': [0.5 + 0.32 * math.sin(i * 0.75) + 0.12 * math.cos(i * 1.7) for i in range(30)],
}, key='id')

with hm.Scene() as scene:
    with hm.Row(wrap=True, min_width=320):
        with hm.Plot(data=observations, height=350, axes2d=True) as source:
            hm.Points(x='x', y='y', size=6, fill='#547ba2',
                             select=True, hover=True, brush=True, respond=True)
        with hm.Plot(data=observations, height=350, axes2d=True):
            with hm.Filter(where='@selected', inactive='all'):
                hm.Points(x='x', y='y', size=7.5, fill='#cf8955')
scene
```

```python
# Programmatic changes use the same data-owned selection resource.
await source.selection.replace([0, 3, 5])
```

```python
# Selections survive a compatible keyed table replacement.
updated_y = [0.5 + 0.32 * math.sin(i * 0.75 + 0.3) + 0.12 * math.cos(i * 1.7) for i in range(30)]
observations.update_columns(y=updated_y)
```
