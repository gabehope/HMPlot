# Shared and independent scales
A normalizer maps source values to a common unit interval. Color and size encode that interval independently. Same-named columns in separate tables remain independent unless given the same scale resource.

```python
import hmplot as hm

observations = {'x': [0, 1, 2, 3, 4], 'y': [2, 4, 3, 5, 4], 'score': [8, 25, 42, 78, 96], 'group': ['Low', 'Low', 'Middle', 'High', 'High']}
comparison = {'x': [0, 1, 2, 3, 4], 'y': [5, 4, 3, 2, 1], 'score': [20, 35, 40, 60, 82]}
shared = hm.normalization(domain=[0, 100])
color = hm.continuous(normalization=shared, colormap='Viridis')
size = hm.continuous(normalization=shared, range=[8, 22])
ordered = hm.ordinal(domain=['Low', 'Middle', 'High'], colormap='Viridis')

with hm.Scene() as scene:
    with hm.Row(wrap=True, min_width=320):
        with hm.Plot(data=observations, axes2d=True, height=320):
            points = hm.Points(x='x', y='y', fill='score', size='score', fill_scale=color, size_scale=size)
            hm.Legend(target=points, channel='fill', label='Score')
        with hm.Plot(data=comparison, axes2d=True, height=320):
            hm.Points(x='x', y='y', fill='score', size=6)
        with hm.Plot(data=observations, axes2d=True, height=320):
            groups = hm.Points(x='x', y='y', fill='group', fill_scale=ordered, size=7.5)
            hm.Legend(target=groups, channel='fill', label='Band')
scene
```
