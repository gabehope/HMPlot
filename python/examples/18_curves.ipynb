# Curves and billboarded arrows


```python
import hmplot as hm
with hm.Plot() as plot:
    with hm.Frame(x={"domain": [-1, 5]}, y={"domain": [-1, 4]}):
        hm.Line(position=[[0, 0], [1, 2], [2, 1], [3, 3], [4, 2]], curve="catmull-rom", width=2)
        hm.charts.Arrows(start=[[0, -0.5]], end=[[4, -0.5]], curve="cubic", control1=[[1, 1]], control2=[[3, 1]])
plot
```

```python
hm.visualize_tree(plot)
```
