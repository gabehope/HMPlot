# Density


```python
import hmplot as hm
import numpy as np

values = hm.Table({"group": ["A", "B"] * 120, "value": np.sin(np.arange(240) * 1.7) + np.tile([-1, 1], 120)})
with hm.Plot(data=values, axes2d=True, height=420) as plot:
    with hm.Density(x="value", by="group", resolution=256):
        hm.Line(x="x", y="density", color="group", width=2)
plot
```
