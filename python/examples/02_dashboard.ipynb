# A composed dashboard


```python
import hmplot as hm
import numpy as np

x = np.linspace(-3, 3, 120)
table = hm.Table({"x": x, "sine": np.sin(x), "cosine": np.cos(x)})
with hm.Scene() as scene:
    with hm.Column(gap=12):
        # Inputs send their values up to the Scene, and everything in it reads them by name:
        # "$series" where a string names a column, hm.input.bins anywhere.
        with hm.Row(wrap=True, min_width=0):
            hm.Select(name="series", options=["sine", "cosine"], label="Series")
            hm.Slider(name="bins", min=4, max=48, step=1, value=16, label="Bins")
        with hm.Data(table):
            with hm.Row(wrap=True, min_width=320):
                with hm.Plot(title="Signal", axes2d=True) as signal:
                    hm.Line(x="x", y="$series", color="steelblue")
                with hm.Plot(title="Histogram", axes2d=True):
                    with hm.stats.Bin(x="$series", bins=hm.input.bins):
                        hm.Rect(x0="x0", x1="x1", y0=0, y1="count")
            hm.Minimap(target=signal, width=600, height=90, axes={"x": True})
scene
```

```python
await scene.export_png("dashboard.png", scale=2)
```
