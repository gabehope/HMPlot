# Interactive force layouts


```python
import hmplot as hm
import numpy as np

edges = {"source": ["A", "A", "B", "C", "D"], "target": ["B", "C", "D", "D", "E"]}
with hm.Scene() as scene:
    # running= follows the checkbox; reheat= acts on each press (a Button's value counts presses).
    with hm.Row(wrap=True, min_width=0):
        hm.Checkbox(name="run", value=True, label="Run")
        hm.Button(name="reheat", label="Reheat")
        hm.Slider(name="repulsion", min=0, max=200, step=1, value=30, label="Repulsion")
    with hm.Data(edges):
        with hm.GraphData(source="source", target="target"):
            with hm.Force(seed=42, running=hm.input.run, reheat=hm.input.reheat, repulsion=hm.input.repulsion) as layout:
                with hm.Plot(axes2d=True) as plot:
                    with hm.Edges():
                        hm.Segments(start="start", end="end", width=0.5)
                    with hm.Nodes(drag=True):
                        hm.Points(position="position", fill="degree", size=7)
scene
```
