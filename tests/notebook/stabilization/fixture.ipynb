```python
import hmplot as hm
import numpy as np
print('HEATMAP_VERSION', hm.__version__)
x = np.linspace(0, 6, 24, dtype=np.float32)
table = hm.Table({'x': x, 'temperature': np.sin(x), 'rain': np.cos(x)*10})
with hm.Scene() as scene:
    with hm.View(data=table, height=280, overview=True) as view:
        with hm.Frame(x={'domain':[0,6]}, y={'domain':[-1,1], 'label':'Temperature'}):
            points = hm.Points(x='x', y='temperature', size=8)
        with hm.Frame(x={'domain':[0,6], 'visible':False}, y={'domain':[-10,10], 'label':'Rain', 'side':'right'}):
            hm.Line(x='x', y='rain', color='coral')
    with hm.Plot(data=table, height=220, axes2d=True):
        histogram = hm.charts.Histogram(x='temperature', bins=8)
scene
```
