# Notebook bridge reproduction
Run in VS Code, JupyterLab, or Notebook 7 from a freshly installed wheel.


```python
import hmplot as hm
import numpy as np
table = hm.Table({'id': np.arange(32), 'x': np.linspace(0, 1, 32), 'y': np.sin(np.linspace(0, 6, 32))}, key='id')
with hm.Scene() as scene:
    with hm.Row():
        with hm.Plot(data=table, title='Interact') as left:
            hm.Points(x='x', y='y', size=10, hover=True, select=True, brush=True, respond=True)
        with hm.Plot(data=table, title='Selected rows'):
            with hm.Filter(where='@selected', inactive='all'):
                hm.Points(x='x', y='y', size=10, fill='#cf8955')
scene
```

```python
display(scene)  # Mount a second independent notebook view.
```

```python
table.update_columns(y=np.cos(np.linspace(0, 6, 32)))
view = next(iter(scene.views))
await scene.when_settled(view=view, timeout=30)
await left.export_png('/tmp/bridge-plot.png', view=view, timeout=30)
await scene.export_svg('/tmp/bridge-scene.svg', view=view, timeout=30)
await scene.export_pdf('/tmp/bridge-scene.pdf', view=view, timeout=30)
print('BRIDGE_OK')
```
