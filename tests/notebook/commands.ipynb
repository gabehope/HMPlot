```python
import hmplot as hm
from IPython.display import display
with hm.Scene() as scene:
    with hm.Row():
        with hm.Plot(title="A", animation_duration=250):
            points = hm.Points(position=[[0, 0], [1, 1]], size=12, animation=True)
        with hm.Plot(title="B"):
            hm.Line(position=[[0, 0], [1, 1]])
display(scene)
```

```python
import asyncio
await asyncio.wait_for(scene.when_animations_finished(), 15)
points.update(size=16)
await asyncio.wait_for(scene.when_settled(), 15)
png = await asyncio.wait_for(scene.export_png(scale=1), 20)
svg = await asyncio.wait_for(scene.export_svg(), 20)
pdf = await asyncio.wait_for(scene.export_pdf(), 20)
assert png.startswith(b"\x89PNG")
assert svg.startswith(b"<svg")
assert pdf.startswith(b"%PDF")
print("COMMANDS_OK", len(png), len(svg), len(pdf))
```
